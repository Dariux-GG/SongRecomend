```python
import pandas as pd

# Cargar los DataFrames
df1_path = "C:/Users/angel/OneDrive/Escritorio/PersonalP/bases de df/df_true.csv"
df2_path = "C:/Users/angel/OneDrive/Escritorio/PersonalP/bases de df/dataset.csv"

df1 = pd.read_csv(df1_path)
df2 = pd.read_csv(df2_path)


print(df1.columns)
print(df2.columns)
```

```python
# 1. Eliminar columnas extras de df2
columnas_a_eliminar = ['Unnamed: 0', 'track_id', 'album_name', 'explicit', 'time_signature', 'track_genre']
df2 = df2.drop(columns=columnas_a_eliminar)

# 2. Renombrar columnas para que coincidan con df1
df2 = df2.rename(columns={
    'track_name': 'song_name',
    'artists': 'artist',
    'duration_ms': 'duration'  # Mantenemos el nombre 'duration' aunque esté en ms
})


orden_columnas = ['song_name', 'artist', 'popularity', 'danceability', 'energy', 'key',
                 'loudness', 'mode', 'speechiness', 'acousticness', 'instrumentalness',
                 'liveness', 'valence', 'tempo', 'duration']

# Reordenar las columnas de df2 para que coincidan con df1
df2 = df2[orden_columnas]

# Verificación final
print(df2.columns)
print(df1.columns)


# Columnas numéricas donde podrían aparecer notación científica (ajusta según tu df2)
columnas_numericas = ['popularity', 'danceability', 'energy', 'loudness', 
                     'speechiness', 'acousticness', 'instrumentalness', 
                     'liveness', 'valence', 'tempo']

# Convertir a float con precisión extendida (evita notación científica)
df2[columnas_numericas] = df2[columnas_numericas].astype('float64')  # o 'float128' si es necesario

# Opcional: Redondear decimales si son muy largos
df2[columnas_numericas] = df2[columnas_numericas].round(10)  # 10 decimales



df2.to_csv('C:/Users/angel/OneDrive/Escritorio/PersonalP/bases de df/datasetclean.csv', index=False)
```
