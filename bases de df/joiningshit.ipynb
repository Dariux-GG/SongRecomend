```python
import pandas as pd

# Cargar los DataFrames
df1_path = "C:/Users/angel/OneDrive/Escritorio/PersonalP/bases de df/df_true.csv"
df2_path = "C:/Users/angel/OneDrive/Escritorio/PersonalP/bases de df/datasetclean.csv"

df1 = pd.read_csv(df1_path)
df2 = pd.read_csv(df2_path)
print(df1.columns)
print(df2.columns)
df2= df2.dropna()
df2.isna().sum()


```

```python
# Juntar y contar antes de eliminar duplicados
df = pd.concat([df1, df2], ignore_index=True)
total_antes = len(df)

# Eliminar duplicados por song_name + artist
df = df.drop_duplicates(subset=["song_name", "artist"])
total_despues = len(df)

# Calcular cuántas coincidencias fueron eliminadas
coincidencias = total_antes - total_despues
print(f"🧹 Canciones duplicadas eliminadas: {coincidencias}")


# Mostrar cuántos duplicados hay
duplicados = df.duplicated(subset=["song_name", "artist"]).sum()
print(f"🔁 Duplicados encontrados: {duplicados}")

# Eliminar duplicados
df_limpio = df.drop_duplicates(subset=["song_name", "artist"])

# Guardar el resultado
df_limpio.to_csv("df_limpio.csv", index=False)
print(f"✅ Archivo limpio guardado como 'df_limpio.csv'")

# Mostrar cuántas canciones únicas quedaron
print(f"🎵 Canciones únicas: {len(df_limpio)}")


df_limpio.to_csv("solo_canciones_artistas.csv", index=False)
```
